**2. Problem Statement**

Diabetes is a common health condition that can affect a person's overall health and may lead to various complications. Early prediction can help identify individuals who may be at risk.

The objective of this project is to develop a machine learning model that can predict whether a person is diabetic or non-diabetic using medical information such as pregnancy count, plasma glucose concentration, blood pressure, skin thickness, insulin level, BMI, diabetes pedigree function, and age.

The model uses Logistic Regression to classify the patients into two categories:

0 → Non-Diabetic
1 → Diabetic

```python
from sklearn.model_selection import train_test_split
x = df.drop('class', axis =1)
y = df['class']
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.3,random_state=1)
```

# MEAN THE ZERO VALUES

```python
from sklearn.impute import SimpleImputer
rep_o = SimpleImputer(missing_values = 0, strategy = 'mean')
colm = x_train.columns
x_train = pd.DataFrame(rep_o.fit_transform(x_train), columns = colm)
x_test = pd.DataFrame(rep_o.transform(x_test), columns = colm)
print(x_train.head())
```

```python
from sklearn import metrics
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(solver = 'liblinear')
model.fit(x_train,y_train)
y_predict = model.predict(x_test)

coef_df = pd.DataFrame(model.coef_, columns = colm)
coef_df['intercept'] = model.intercept_
coef_df.index = ['coefficient']
(coef_df.round(4))
```

```python
model_score = model.score(x_test, y_test)
print(f"Model accuracy : {model_score:.4f}, {model_score*100:.2f}%")
```

```python
cm = metrics.confusion_matrix(y_test,y_predict,labels = (1,0))
cm
```
